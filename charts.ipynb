```python
#Load Libraries for Data Analysis
import pandas as pd
import os
os.listdir('/kaggle/input/datasets/corneliusozoji/stock-price-new')
```

```python
data = pd.read_csv("/kaggle/input/datasets/corneliusozoji/stock-price-new/Stock_Prices.csv")
data.head(5)
```

```python
# Check for missing values (True/False)
data.isnull()
```

```python
# Counting missing values per column
data.isnull().sum()
```

```python
# Total missing values
data.isnull().sum().sum()
```

```python
# Check How Many Records There Are 
data.shape
```

```python
# Drop Rows With Missing Values Since Not Much i.e 27
data_clean = data.dropna()
```

```python
# Check to see output 
data_clean.isnull().sum()
```

```python
# Check How Many Records Remain
data_clean.shape
```

```python
# Check for Duplicates - Shows True for Duplicate Rows
data_clean.duplicated()
```

```python
# Check for No of Duplicates 
data_clean.duplicated().sum()
```

```python
# Display first 10 Records
data_clean.head(10)
```

```python
#Standardize Date Records
data_clean.loc[:, 'date'] = pd.to_datetime(data_clean['date'])
```

```python
# Display first 10 Records
data_clean.head(10)
```

```python
# Summary statistics
data_clean.describe()
```

```python
# Visualaztions- Ploting Bar Charts
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 5))
plt.hist(data_clean['open'], bins=10, alpha=0.5, label='DEBT', color='blue', edgecolor='black')
plt.hist(data_clean['close'], bins=10, alpha=0.5, label='SIZE', color='orange', edgecolor='black')
plt.title('Histogram of Open and Close')
plt.xlabel('Open')
plt.ylabel('Close')
```

```python
# Visualization - Box Plot
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.boxplot(
    [data_clean['open'], data_clean['close']],
    labels=['Open', 'Close'],
    patch_artist=True
)

plt.title('Box Plot of Open and Close')
plt.ylabel('Values')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```

```python
# Visualization - Scatter Plot
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.scatter(
    data_clean['open'],
    data_clean['close'],
    color='green',
    alpha=0.7,
    edgecolor='black'
)
```

```python
# Display first 10 records
print(data_clean.head(10))

# Find correlations between numerical features
correlation_matrix = data_clean.corr(numeric_only=True)

# Display the correlation matrix
print(correlation_matrix)
```
